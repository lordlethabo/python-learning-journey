```python
# Operations
fruits.add("mango")
print(fruits)

# Remove an item
fruits.discard("apple") # Changed from remove to discard to prevent KeyError
print(fruits)

fruits.discard("kiwi")
print(fruits)

# Check membership
print("orange" in fruits)
print("kiwi" in fruits)

# Set operations
a = {1, 2, 3, 4}
b = {3, 4, 5, 6}
print(a.union(b)) # Combine: {1, 2, 3, 4, 5, 6}
print(a.intersection(b)) # Common elements: {3, 4}
print(a.difference(b)) # Items in a not in b: {1, 2}
print(a.symmetric_difference(b)) # In a or b, but not both: {1, 2, 5, 6}
```

```python
# crops
crops_north = {"maize", "wheat", "rice"}
crops_south = {"rice", "sorghum", "maize"}

# Crops grown in either north or south
all_crops = crops_north.union(crops_south)
print("All crops:", all_crops)

# Crops grown in both north and south
common_crops = crops_north.intersection(crops_south)
print("Common crops:", common_crops)

# Crops only grown in north
only_north = crops_north.difference(crops_south)
print("Only north crops:", only_north)
```

```python
# Set operations
A = {1, 2, 3, 4}
B = {3, 4, 5, 6}

print("Union: ", A.union(B)) # {1, 2, 3, 4, 5, 6}
print("Intersectin: ", A.intersection(B)) # ({3, 4})
print("Difference: ", A.difference(B)) # {1, 2}
print("Symmetric_difference: ", A.symmetric_difference(B)) # {1, 2, 5, 6}
```

```python
#mini exercise 2
north_crops = {"maize", "wheat", "rice"}
south_crops = {"rice", "sorghum", "maize"}

# Crops in either north or south
all_crops = north_crops.union(south_crops)
print("All crops: ", all_crops)

# Crops in both regions
common_crops = north_crops.intersection(south_crops)
print("Common crops: ", common_crops)

# Crops only in north
only_north = north_crops.difference(south_crops)
print("Only in the north: ", only_north)
```
